```python
from pathlib import Path
import sys
import wfdb
import matplotlib.pyplot as plt

project_root = Path.cwd()
if not (project_root / "src" / "data_load.py").exists():
    project_root = project_root.parent
sys.path.insert(0, str(project_root))

from src.data_load import  load_noisy_ecg_data
```

```python

ecg_path = r"C:\Users\Luchitha\Documents\Python\EMD_for_ECG\data\physionet.org\files\mitdb\1.0.0\100"
multiple_noises = [ "ma" , "em"]  # Change this to "ma" or "em" for different noise types
```

```python
snr = 10

clean_ecg_signal, noise_signal, noisy_ecg_signal, fs = load_noisy_ecg_data(ecg_path, snr , multiple_noises=multiple_noises)


```

```python
# Plot the clean , noise , and noisy ECG signals
plt.figure(figsize=(15, 10))
plt.subplot(3, 1, 1)
plt.plot(clean_ecg_signal[:5000], color='blue')
plt.title('Clean ECG Signal')
plt.xlabel('Sample')
plt.ylabel('Amplitude')
plt.subplot(3, 1, 2)
plt.plot(noise_signal[:5000], color='red')
plt.title(f'Noise Signal ({", ".join(n + " Noise" for n in multiple_noises)})')
plt.xlabel('Sample')
plt.ylabel('Amplitude')
plt.subplot(3, 1, 3)
plt.plot(noisy_ecg_signal[:5000], color='green')
plt.title(f'Noisy ECG Signal (SNR = {snr} dB)')
plt.xlabel('Sample')
plt.ylabel('Amplitude')
plt.tight_layout()
plt.show()
```
